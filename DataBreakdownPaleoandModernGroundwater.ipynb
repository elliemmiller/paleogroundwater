## Modern Exfiltration Data Breakdown

```python
import pandas as pd

# Load the CSV file
file_path = 'modernexfiltrationperm11.csv'  # your CSV file
df = pd.read_csv(file_path)

# Extract the exfiltration column
exfil_col = df['exfiltration_mm_per_yr']

# Calculate mean, max, min
mean_exfil = exfil_col.mean()
max_exfil = exfil_col.max()
min_exfil = exfil_col.min()

# Print results
print(f"Mean exfiltration (mm/yr): {mean_exfil}")
print(f"Maximum exfiltration (mm/yr): {max_exfil}")
print(f"Minimum exfiltration (mm/yr): {min_exfil}")
```
