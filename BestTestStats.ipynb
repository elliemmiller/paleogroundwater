```python
import pandas as pd

# Read data
df = pd.read_excel("BestSensitivityTest/besttestssummary.xlsx", sheet_name="Sheet1")

# Get r² for each site's selected configuration
df["selected_r2"] = [
    row[row["SensitivityTest"]]
    for _, row in df.iterrows()
]

# Count configurations
counts = df["SensitivityTest"].value_counts()

# Apply r² > 0.5 cutoff
included = df[df["selected_r2"] > 0.5]
excluded = df[df["selected_r2"] <= 0.5]

print(f"Total candidate sites: {len(df)}")
print(f"Included sites: {len(included)}")
print(f"Excluded sites: {len(excluded)}")

print("\nConfiguration counts:")
print(counts)

print("\nConfiguration percentages:")
print((counts / len(df) * 100).round(1))

print("\nExcluded sites:")
print(excluded["Group"].tolist())
```
